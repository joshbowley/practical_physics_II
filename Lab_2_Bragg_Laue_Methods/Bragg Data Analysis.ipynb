# Angle uncertainty, count rate uncertainty and error bars 

Calculating uncertainty values and adding error bars to graphs. 

## Angle uncertainty 

Angle uncertainty is fixed, and given by the resolution of the X-ray unit. 

Since step count is $\theta_{step} = 0.1^\circ$, then angular uncertainty is: 

$$ 
\delta_{angular} = \pm 0.05^\circ 
$$ 
## Count rate uncertainty 

A Geiger-Müller tube can be modelled as a Poisson distrebution when there are enough data points [1]. Thus uncertainties in the measured count rate can be assumed to follow Poisson statistics. 

For a Poisson proces with mean count $N$: 

$$
P(n) = \frac{N^n e^{-N}}{n!}
$$

Thus variance is (see page * in lab book for derivation): 

$$
\text{variance} = \delta ^2 = N
$$

Therefore:

$$
\delta_N = \sqrt{N}
$$

## Converting to count rate

$N$ counts are recorded in time $T$. 'Count rate' is another way of saying intensity $I$ over time period $T$. 

Then: 

$$
I = \frac{N}{T} 
\qquad
\qquad
\delta_I = \frac{\sqrt{N}}{T}
$$

Gate time $T = 2s$. 

Therefore: 

$$
\delta_I = \sqrt{ \frac{I}{2} }
$$

### Reference

[1] University of California, Davis, *Counting Statistics*.  
https://122.physics.ucdavis.edu/sites/default/files/files/Nuclear%20Decay/Counting%20Statistics%20122.pdf

## Error Bars: NaCl (100) first reading

```python
# Angular uncertainty
NaCl_100df["delta_theta"] = 0.05
```

```python
# Intensity uncertainty computation
NaCl_100df["delta_intensity"] = np.sqrt(NaCl_100df["count_rate"] / 2)
```

```python
display(NaCl_100df.head())
```

```python
plt.figure(figsize=(8,5))
plt.plot(NaCl_100df["theta_deg"], NaCl_100df["count_rate"], 'o-', markersize=3)
plt.xlabel(r"Crystal angle $\theta$ (deg)")
plt.ylabel("Count rate (s$^{-1}$)")
plt.title("NaCl (100) Bragg scan: First Measurement")
plt.grid(alpha=0.3)

plt.errorbar(
    NaCl_100df["theta_deg"],
    NaCl_100df["count_rate"],
    xerr=NaCl_100df["delta_theta"],
    yerr=NaCl_100df["delta_intensity"],
    fmt=".",
    capsize=1
)

plt.show()
```

This is hard to read with the $\theta$ error bars plotted. Since $\delta_{\theta}$ is constant for all data points and significantly smaller than the angular range of the scan. Plotting angular error bars therefore adds no additional information and obscures the identification of Bragg peaks. The uncertainty in $\theta$ is instead included explicitly in the calculation of interplanar spacings

```python
plt.figure(figsize=(8,5))
plt.plot(NaCl_100df["theta_deg"], NaCl_100df["count_rate"], 'o-', markersize=3)
plt.xlabel(r"Crystal angle $\theta$ (deg)")
plt.ylabel("Count rate (s$^{-1}$)")
plt.title("NaCl (100) Bragg scan: First Measurement")
plt.grid(alpha=0.3)

plt.errorbar(
    NaCl_100df["theta_deg"],
    NaCl_100df["count_rate"],
    yerr=NaCl_100df["delta_intensity"],
    fmt=".",
    capsize=1
)

plt.show()
```

This is still hard to read and not very useful. The uncertainty varies strongly with intensity and plotting error bars on every data point obscures the overall structure of the scan and the identification of Bragg peaks.

Since the primary purpose of these plots is qualitative peak identification rather than quantitative fitting, I am going to omit error bars for clarity. The relevant uncertainties were instead included explicitly in the numerical determination of Bragg angles and interplanar spacings.
