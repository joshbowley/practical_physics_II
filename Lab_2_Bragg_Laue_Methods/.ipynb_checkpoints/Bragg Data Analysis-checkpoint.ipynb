# Bragg Method Data analysis

Plotting graphs for peak identification in Bragg method, as well as other stuff. 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Graph Visualisation  NaCl (100) first measurement:

```python
# Read into pandas DataFrame
NaCl_100df = pd.read_csv(
    r"C:\Users\joshb\OneDrive\PHYSICS BSc UNIVERSITY of EXETER\BSc Physics Stage 2\Practical Physics II\practical_physics_II\Lab_2_Bragg_Laue_Methods\Bragg Measurements\Textfile nacl 110 first measurement",
    sep="\t",
    encoding="latin1"
)
```

```python
# Rename columns
NaCl_100df.columns = ["theta_deg", "counts"]
```

```python
display(NaCl_100df.head())
print("Data shape:", NaCl_100df.shape)
```

```python
plt.figure(figsize=(8,5))
plt.plot(NaCl_100df["theta_deg"], NaCl_100df["counts"], 'o-', markersize=3)
plt.xlabel(r"Crystal angle $\theta$ (deg)")
plt.ylabel("Count rate (s$^{-1}$)")
plt.title("NaCl (100) Bragg scan")
plt.grid(alpha=0.3)
plt.show()
```

^ Ok, this looks like nonsense. The main issue is row 0 is not data. It is the second header line from the Measure export, which pandas has tried to treat as a data row. 

This means that theta_deg is an object column, not float. This means matplotlib is treating theta as categorical, causing the weird plots. 

```python
# Drop the first row
NaCl_100df = NaCl_100df.drop(index=0).reset_index(drop=True)
```

```python
# Convert data type of both columns to numeric
NaCl_100df["theta_deg"] = pd.to_numeric(NaCl_100df["theta_deg"])
NaCl_100df["counts"] = pd.to_numeric(NaCl_100df["counts"])
```

```python
# Sanity check
NaCl_100df.dtypes
NaCl_100df.head()
```

```python
plt.figure(figsize=(8,5))
plt.plot(NaCl_100df["theta_deg"], NaCl_100df["counts"], 'o-', markersize=3)
plt.xlabel(r"Crystal angle $\theta$ (deg)")
plt.ylabel("Count rate (s$^{-1}$)")
plt.title("NaCl (100) Bragg scan: First Measurement")
plt.grid(alpha=0.3)
plt.show()
```

## Graph Visualisation  NaCl (111) first measurement: 

```python
# Read into pandas DataFrame
NaCl_111df = pd.read_csv(
    r"C:\Users\joshb\OneDrive\PHYSICS BSc UNIVERSITY of EXETER\BSc Physics Stage 2\Practical Physics II\practical_physics_II\Lab_2_Bragg_Laue_Methods\Bragg Measurements\NaCl 111 measurement 1.txt",
    sep="\t",
    encoding="latin1"
)
```

```python
# Rename columns
NaCl_111df.columns = ["theta_deg", "counts"]
```

```python
display(NaCl_111df.head())
print("Data shape:", NaCl_111df.shape)
```

No need to change data or drop first row. 

```python
plt.figure(figsize=(8,5))
plt.plot(NaCl_111df["theta_deg"], NaCl_111df["counts"], 'o-', markersize=3)
plt.xlabel(r"Crystal angle $\theta$ (deg)")
plt.ylabel("Count rate (s$^{-1}$)")
plt.title("NaCl (111) Bragg scan: First Measurement")
plt.grid(alpha=0.3)
plt.show()
```

This plot is dominated by high values at $0^\circ \le \theta \le 4^\circ$. This makes it hard to identify angles of peak intensity. 

I think the reason for the high intensity at the start is the machine held the crystal at too low an angle initially and thus a straight through reading was picked up with the X-Rays firing straight onto the tube. 

This will introduce zero error with the crystal being held behind a few degrees for the whole reading. 

This is not an issue as the ratio between peaks is important, not the peaks themselves. 

To make it easier to plot angles of peak intensity, I will replot without first values affecting scaling. 

```python
theta_min = 5.0  # degrees

roi = NaCl_111df[NaCl_111df["theta_deg"] >= theta_min]

plt.figure(figsize=(8,5))
plt.plot(roi["theta_deg"], roi["counts"], 'o-', markersize=3)
plt.xlabel(r"Crystal angle $\theta$ (deg)")
plt.ylabel(r"Count rate (s$^{-1}$)")
plt.title("NaCl (111) Bragg scan (θ ≥ 5°)")
plt.grid(alpha=0.3)
plt.show()
```
